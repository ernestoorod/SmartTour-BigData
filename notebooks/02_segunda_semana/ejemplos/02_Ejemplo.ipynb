```python
grupo1 = ("Juan", "Carla", "Sebastian", "Maria", "Lucia", "Carla")
grupo2 = ("Juan", "Javier", "Laura", "Paula", "Carlos")

print(f"Repetidas por no ser conjunto {grupo1}")

grupo1 = set(grupo1)
grupo2 = set(grupo2)

print(f"No repetidas por ser conjunto: {grupo1}")

union = grupo1 | grupo2
interseccion = grupo1 & grupo2
diferencia = grupo1 - grupo2
diferencia_simetrica = grupo1 ^ grupo2

print(f"\nGrupo 1: {grupo1}")
print(f"Grupo 2: {grupo2}")
print(f"\nUnion: {union}\nInterseccion: {interseccion}\nDiferencia: {diferencia}\nDiferencia Simetrica: {diferencia_simetrica}")
```
