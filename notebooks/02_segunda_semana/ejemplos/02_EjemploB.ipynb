```python
import csv

grupo1 = set()
grupo2 = set()

with open("grupo1.csv", encoding="utf-8") as f:
    lector = csv.reader(f)
    next(lector)
    
    for fila in lector:
        grupo1.add(fila[0])

with open("grupo2.csv", encoding="utf-8") as f:
    lector = csv.reader(f)
    next(lector)

    for fila in lector:
        grupo2.add(fila[0])

print(f"Grupo 1: {grupo1}\nGrupo 2: {grupo2}")
```
